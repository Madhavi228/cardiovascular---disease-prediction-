# **Project Summary -**
The cardiovascular risk dataset contains information about individuals and their health-related factors that may influence the likelihood of developing coronary heart disease (CHD) within ten years. The dataset consists of 3,390 records and 17 variables, including demographic details, lifestyle habits, medical history, and clinical measurements. The main target variable is TenYearCHD, which indicates whether an individual is predicted to develop coronary heart disease within a ten-year period.

The dataset includes demographic information such as age, sex, and education level. The average age of participants is approximately 49.5 years, with individuals ranging from 32 to 70 years old. The dataset contains slightly more females (1,923) than males (1,467). Education level is represented numerically and provides an indication of socioeconomic background, which can be an important factor in health outcomes.

Lifestyle-related variables include smoking status and the number of cigarettes smoked per day. Smoking is an important cardiovascular risk factor, and the dataset contains almost equal numbers of smokers and non-smokers. Medical and behavioral factors such as blood pressure medication use, history of stroke, hypertension, and diabetes are also included. These variables help identify individuals with existing conditions that may increase their risk of future heart disease.

The clinical measurements in the dataset include total cholesterol, systolic blood pressure, diastolic blood pressure, body mass index (BMI), heart rate, and glucose levels. The average systolic blood pressure is around 132.6 mmHg, while the average BMI is approximately 25.8, suggesting that many individuals fall near the overweight category. The mean total cholesterol level is about 237 mg/dL, which may indicate elevated cholesterol levels for a significant portion of the population. Glucose levels are also included as an indicator of metabolic health.

The target variable shows that most individuals in the dataset do not develop CHD within ten years. Out of the 3,390 participants, 511 individuals (about 15%) are classified as having a positive CHD outcome, while 2,879 individuals (about 85%) are classified as negative. This imbalance is common in medical prediction datasets, where disease cases are usually fewer than healthy cases. It is an important consideration when building machine learning models because accuracy alone may not provide a complete evaluation of performance.

Some missing values are present in the dataset. The largest amount of missing information occurs in glucose measurements, where 304 records lack data. Other variables such as education, blood pressure medication use, cholesterol, cigarettes per day, and BMI also contain smaller numbers of missing entries. These missing values would need to be handled through techniques such as data imputation or removal of incomplete records before analysis or modeling.

Overall, this dataset provides a useful collection of cardiovascular risk indicators. It combines lifestyle, demographic, and medical information to support analysis of factors associated with coronary heart disease. The dataset can be used for statistical analysis, risk factor identification, and machine learning approaches aimed at predicting future cardiovascular events. Understanding relationships between variables such as age, smoking, blood pressure, cholesterol, and diabetes can help in developing strategies for early detection and prevention of heart disease.

#### Chart - 3

````python
# Chart - 3 visualization code
for col in numeric_features[1:-1]:
    fig = plt.figure(figsize=(9, 3))
    ax = fig.gca()

    # Drop NaNs for the current feature and the target variable together
    temp_df = df[[col, 'TenYearCHD']].dropna()
    feature_clean = temp_df[col]
    label_clean = temp_df['TenYearCHD']

    correlation = feature_clean.corr(label_clean)

    plt.scatter(x=feature_clean, y=label_clean)
    plt.xlabel(col)
    plt.ylabel('TenYearCHD')
    ax.set_title(f'TenYearCHD vs {col} - correlation: {correlation:.4f}')

    # Ensure there's enough data to fit a line and sufficient variance
    if feature_clean.nunique() > 1 and len(feature_clean) > 1 and feature_clean.std() > 1e-9:
        try:
            # Set rcond to a small value to improve numerical stability for polyfit
            z = np.polyfit(feature_clean, label_clean, 1, rcond=1e-6)
            y_hat = np.poly1d(z)(feature_clean)
            plt.plot(feature_clean, y_hat, "r--", lw=1)
        except np.linalg.LinAlgError as e:
            print(f"Skipping linear fit for {col} due to LinAlgError (even with rcond): {e}")
            print(f"Details: {col} unique values: {feature_clean.nunique()}, std: {feature_clean.std()}, data points: {len(feature_clean)}")
    else:
        print(f"Skipping linear fit for {col} as there are not enough unique values or sufficient variance.")

plt.show()
````

## ***7. ML Model Implementation***

### ML Model - 1

````python
# ML Model - 1 Implementation
model = DecisionTreeClassifier(max_depth=10,
    random_state=42)

# Fit the Algorithm
model.fit(x_train,y_train)


# Predict on the model
train_pred = model.predict(x_train)
test_pred = model.predict(x_test)



````

````python
train_pred = model.predict(x_train)
train_pred = np.where(train_pred>0.5,1,0)
train_pred
````

````python
from sklearn.metrics import confusion_matrix,accuracy_score,precision_score,recall_score
print('Train Confusion Matrix \n',confusion_matrix(y_train,train_pred))
print('Train Accuracy ',accuracy_score(y_train,train_pred))
print('Train Precision ',precision_score(y_train,train_pred))
print('Train Recall ',recall_score(y_train,train_pred))
````

````python
print('Test Confusion Matrix \n', confusion_matrix(y_test,test_pred))
print('Test Accuracy ',accuracy_score(y_test,test_pred))
print('Test Precision ',precision_score(y_test,test_pred))
print('Test Recall ',recall_score(y_test,test_pred))
````
